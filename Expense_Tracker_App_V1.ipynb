# **Step 1 – Collect expenses (get_expenses())**

**Business requirement FR-01:**

Ask the user for item name, amount and category, one expense at a time. Stop when the user types done.

**Business requirement FR-02:**

If the user leaves the category blank, save the
expense as "other" (Default Catagory).


```python
print("Hi! Welcome!")
```

```python
def get_expenses():
  expenses = []

  while True:
    item = input("Item (or 'done' to finish): ")

    if item.lower() == 'done':
      break

    amount = float(input("Amount (TK): "))

    catagory = input("Catagory: ").strip.lower

    if catagory == "":
      catagory == "others"

    expenses.append(expenses)

  return expenses
```

# **Step 2 – Calculate category totals**

**Business requirement FR-03:**

 Group all expenses by category and add up each category.

```python
def calculate_catagory_totals(expenses):
  catagory_totals = {}

  for expense in expenses:
    catagory = expenses["category"]
    amount = expenses["amount"]

    if catagory in catagory_totals:
      catagory_totals[catagory] += amount
    else:
      catagory_totals[catagory] = amount

  return catagory_totals
```

# **Step 3 – Calculate overall spending**
**Business requirement FR-04:**

Show the total amount spent in the month.

```python
def calculate_total(expenses):
  total = 0

  for expense in expenses:
    total += expenses["amount"]

  return total
```
