# **Expense Tracker App**
Code Language: Python

# **Step 1 – Collect expenses (get_expenses())**

**Business requirement FR-01:**

Ask the user for item name, amount and category, one expense at a time. Stop when the user types done.

**Business requirement FR-02:**

If the user leaves the category blank, save the
expense as "other" (Default Category).


```python
print("Hi! Welcome!")
```

```python
def get_expenses():

    expenses = []

    while True:

        item = input("Item (or 'done' to finish): ")

        if item.lower() == 'done':
            break

        amount = float(input("Amount (Tk): "))

        category = input("Category: ").strip().lower()

        if category == "":
            category = "other"

        expense = {
            "item": item,
            "amount": amount,
            "category": category
        }

        expenses.append(expense)

    return expenses
```

# **Step 2 – Calculate category totals**

**Business requirement FR-03:**

 Group all expenses by category and add up each category.

```python
def calculate_category_totals(expenses):

    category_totals = {}

    for expense in expenses:

        category = expense["category"]
        amount = expense["amount"]

        if category in category_totals:
            category_totals[category] += amount
        else:
            category_totals[category] = amount

    return category_totals
```

# **Step 3 – Calculate overall spending**
**Business requirement FR-04:**

Show the total amount spent in the month.

```python
def calculate_total(expenses):
  totals = 0

  for expense in expenses:
    totals += expense["amount"]

  return totals
```

# **Step 4 – Find the top spending category**
**Business requirement FR-05:**

 Show the category with the highest spending and its amount.

```python
def find_top_category(category_totals):

  if not category_totals:
    return None, 0

  top_category = max(category_totals, key = category_totals.get)

  top_amount = category_totals[top_category]

  return top_category, top_amount
```

# **Step 5 – Print the monthly report**

**Business requirement FR-06:**

 Print each category, its amount and its share of the total in percentage.

```python
def print_report(category_totals, totals, top_category, top_amount):

    if totals == 0:
        print("No expenses entered.")
        return

    print("-------Monthly Report--------")

    for category, amount in category_totals.items():

        percentage = (amount / totals) * 100

        print(f"{category}: {amount:g} Tk ({percentage:.0f}%)")

    print("--------------------------")
    print(f"Total: {totals:g} Tk")

    print(
        f"You spent the most on {top_category}: "
        f"{top_amount:g} Tk"
    )
```

# **Step 6 – Connect everything using main()**

```python
def main():

    expenses = get_expenses()

    category_totals = calculate_category_totals(expenses)

    total = calculate_total(expenses)

    top_category, top_amount = find_top_category(category_totals)

    print_report(category_totals, total, top_category, top_amount)


if __name__ == "__main__":
    main()
```
